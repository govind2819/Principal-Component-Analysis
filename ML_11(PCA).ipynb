#**PCA (Principal Component Analysis)**



*   it is feature rediction technique.
*   An unsupervised ML technique, used for dimensional reduction.
*   It transforms original numeric feature into new features called PCA.


---

*   How it works ?
1.   Collect the numeric data.
2.   Standardize the feature.
3.   Calculate the principal directions that capture the greatest variance using eigen vector and eigen value.
4.   Select the Components.
5.   Transform the data.








*   Eigen Vector - The direction in which we look at the data.
*   Eigen Value - How much Variance or spread exist in that direction.

                        A.u = λ.V





> PC1 = The direction that captures the maximum variance.

> PC2 = The direction that captures the maximum remaining variance while being perpendicular to PC1.

> PC3 = The next direction perpendicular to the precedding components.

```python
import pandas as pd
df = pd.DataFrame(data)
```

```python
df
```

```python
 X = df.drop('Income',axis =1)
 Y = df['Income']
```

```python
Y
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_scal = sc.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
pca = PCA(
    n_components=0.95
)
```

```python
x_train_pca = pca.fit_transform(x_scal)
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(x_train_pca,Y, test_size = 0.25, random_state = 42)
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
```

```python
model.fit(xtrain,ytrain)
```

```python
y_pred = model.predict(xtest)
```

```python
y_pred
```

#Evaluation

```python
from sklearn.metrics import mean_absolute_error,r2_score
print('Mean Absolute Error : ', mean_absolute_error(ytest,y_pred) )
print('R2 Score : ', r2_score(ytest,y_pred))
```

- R² ≈ 1: Excellent fit; the model explains most of the variance in salary.
- R² ≈ 0: The model performs roughly like predicting the average salary.
- R² < 0: The model performs worse than predicting the average salary.

-  MAE ≈ 0: Excellent predictions; predicted salaries are very close to actual salaries.
